#### FINANCIAL DRIVERS

```python
churn_count = df[df.Churn == "Yes"]
```

```python
sns.histplot(x="MonthlyCharges", data=churn_count, bins=10, color = "purple", stat = "probability")
plt.ylabel("Churn rate")
plt.title("Monthly Charges vs Churn rate")
plt.show()
```

#### Proxy Variable

```python
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"], errors= "coerce")
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
sns.histplot(x="TotalCharges", data=churn_count, bins=9, color = "purple", ax= axes[0])
sns.scatterplot(data= df, x= "tenure", y= "TotalCharges", hue= "Churn", alpha = 0.7, ax= axes[1])
plt.tight_layout()
plt.show()
```

##### Total charges show a steady negative relationship with the number of clients who churned and they also show a strong positive correlation with tenure.
##### Insight: TotalCharges are not a direct driver of retention, but a proxy for tenure and service bundling. Customers with higher charges tend to be long‑tenured and subscribed to multiple services, both of which reduce churn. This explains the steady decline in churn rates as TotalCharges increase.

##### Out of 21 variables, 11 emerged as key churn drivers with rates at least twice as high as their counterparts. While other variables also show elevated churn (~17% vs ~6%), they cluster around similar levels and provide less direct leverage for retention strategy. For clarity and actionability, this report focuses on the most strategic drivers.

```python
corr_matrix = df[["tenure", "MonthlyCharges", "TotalCharges"]].corr()
```

```python
sns.heatmap(corr_matrix, annot=True, cmap="coolwarm", center=0)
```

#####  The heatmap shows that tenure is the strongest determinant of total charges (corr = 0.83). Monthly charges alone have little bearing on how long a client stays (corr = 0.25), but they contribute moderately to total charges (corr = 0.65). This confirms that churn is less about the monthly bill size and more about tenure and bundled services; the longer a customer stays, the more they pay in total, and the less likely they are to churn.
